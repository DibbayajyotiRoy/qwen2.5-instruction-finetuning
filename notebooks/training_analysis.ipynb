# Qwen2.5 Training Analysis

This notebook loads TensorBoard scalars and plots training/validation loss, learning rate, gradient norms, and GPU memory.

```python
from tensorboard.backend.event_processing import event_accumulator
import matplotlib.pyplot as plt
from pathlib import Path

RUN_DIR = Path("outputs/qlora/runs")
event_file = next(RUN_DIR.glob("events.out.tfevents.*"))
ea = event_accumulator.EventAccumulator(str(event_file))
ea.Reload()

print("Available tags:", ea.Tags()["scalars"])
```

```python
def plot_scalar(tag):
    data = ea.Scalars(tag)
    steps = [x.step for x in data]
    values = [x.value for x in data]
    plt.figure(figsize=(10, 4))
    plt.plot(steps, values)
    plt.title(tag)
    plt.xlabel("Step")
    plt.ylabel("Value")
    plt.grid(True)
    plt.show()

plot_scalar("train/loss")
plot_scalar("eval/loss")
plot_scalar("train/learning_rate")
```
